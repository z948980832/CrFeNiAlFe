```python
import pandas as pd
import numpy as np
```

```python
alloy=pd.read_csv('./chaoyu/pre.csv')
alloy=alloy[['Al','Co','Cr','Fe','Ni','Mn','Cu','Nb','Ti','Mo']]*0.01
alloy #元素比
```

```python
element=pd.read_csv('./element.csv')
element.drop(columns = ['Unnamed: 0'],inplace = True)
element.index = ['atom radius(pm)', 'Tm(K)', 'p(kg/m3)','Youngs modulus(GPa)','EN','VEC','bulk modulus(GPa)']
element
```

```python
parameter=pd.DataFrame()#parameter存放最终计算出来的参数
```

```python
#1.理论密度
p1=element.loc['p(kg/m3)']*alloy
parameter['p(kg/m3)'] = p1.apply(lambda x: x.sum(), axis=1) 
```

```python
#1.平均原子半径Mean atom radius
parameter1=element.loc['atom radius(pm)']*alloy
parameter['Mean atom radius'] = parameter1.apply(lambda x: x.sum(), axis=1) 
```

```python
#2.原子尺寸差Atomic size difference
parameter2=pd.DataFrame(columns=element.columns)
for i in range(parameter.shape[0]):
    parameter2.loc[i]=(1-element.loc['atom radius(pm)']/parameter.loc[i, 'Mean atom radius'])**2
parameter2=parameter2*alloy
parameter['Atomic size difference'] = parameter2.apply(lambda x: x.sum()**0.5, axis=1) 
```

```python
#3.平均熔点Average of the melting points
parameter3=element.loc['Tm(K)']*alloy
parameter['Average of the melting points'] = parameter3.apply(lambda x: x.sum(), axis=1) 
```

```python
#4.熔化温度标准差Standard deviation of melting temperature
parameter4=pd.DataFrame(columns=element.columns)
for i in range(parameter.shape[0]):
    parameter4.loc[i]=(1-element.loc['Tm(K)']/parameter.loc[i, 'Average of the melting points'])**2
parameter4=parameter4*alloy
parameter['Standard deviation of melting temperature'] = parameter4.apply(lambda x: x.sum()**0.5, axis=1) 
```

```python
#5.电负性Electronegativity
parameter5=element.loc['EN']*alloy
parameter['Electronegativity'] = parameter5.apply(lambda x: x.sum(), axis=1) 
```

```python
#6.电负性标准差Standard deviation of electronegativity
parameter6=pd.DataFrame(columns=element.columns)
for i in range(parameter.shape[0]):
    parameter6.loc[i]=(element.loc['EN']-parameter.loc[i, 'Electronegativity'])**2
parameter6=parameter6*alloy
parameter['Standard deviation of electronegativity'] = parameter6.apply(lambda x: x.sum()**0.5, axis=1) 
```

```python
#7.Average VEC
parameter7=element.loc['VEC']*alloy
parameter['Average VEC'] = parameter7.apply(lambda x: x.sum(), axis=1) 
```

```python
#8.Standard deviation of VEC
parameter8=pd.DataFrame(columns=element.columns)
for i in range(parameter.shape[0]):
    parameter8.loc[i]=(element.loc['VEC']-parameter.loc[i, 'Average VEC'])**2
parameter8=parameter8*alloy
parameter['Standard deviation of VEC'] = parameter8.apply(lambda x: x.sum()**0.5, axis=1)
```

```python
#9.Mean bulk modulus
parameter9=element.loc['bulk modulus(GPa)']*alloy
parameter['Mean bulk modulus'] = parameter9.apply(lambda x: x.sum(), axis=1) 
```

```python
#10.Standard deviation of bulk modulus
parameter10=pd.DataFrame(columns=element.columns)
for i in range(parameter.shape[0]):
    parameter10.loc[i]=(element.loc['bulk modulus(GPa)']-parameter.loc[i, 'Mean bulk modulus'])**2
parameter10=parameter10*alloy
parameter['Standard deviation of bulk modulus'] = parameter10.apply(lambda x: x.sum()**0.5, axis=1)
```

```python
#11.Mixing entropy
parameter11=alloy*(-alloy.apply(np.log))
parameter['Mixing entropy']=8.314*parameter11.apply(lambda x: x.sum(), axis=1)
```

```python
#12.Mixing enthalpy

#得到所有组合的Hij
Hij=pd.read_csv('./chaoyu/Hij.csv',index_col=0,header=None)
Hij=Hij.T

#alloy列表两两相乘
# 获取列名列表
column_names = alloy.columns.tolist()
# 创建新的DataFrame来存储相乘的结果
parameter12 = pd.DataFrame()
# 对列进行两两相乘
for i in range(len(column_names)):
    for j in range(i + 1, len(column_names)):
        col_i = column_names[i]
        col_j = column_names[j]
        product_col = alloy[col_i] * alloy[col_j]
        new_col_name = f"{col_i}{col_j}"
        product_col=pd.DataFrame(product_col,columns=[new_col_name])
        parameter12=pd.concat([parameter12,product_col],axis=1)
# 提取Hij相同列名的相乘结果
parameter12=parameter12.loc[:,Hij.columns.tolist()]

Hij_data = Hij.iloc[0] #提取Hij的数据
result12 = parameter12.mul(Hij_data, axis=1)#逐行相乘
parameter['Mixing enthalpy'] = result12.apply(lambda x: x.sum()*4, axis=1) #
```

```python
#13.Standard deviation of mixing enthalpy
parameter13 = pd.DataFrame()
for i in range(alloy.shape[0]):
    result13=Hij-parameter.loc[i, ['Mixing enthalpy']].values[0]
    parameter13 = pd.concat([parameter13,result13],axis=0) #Hij减去每条数据的Mixing enthalpy（Hij−ΔHmix）

parameter13=parameter13**2#(Hij−ΔHmix)²
parameter13.index=range(parameter13.shape[0])

parameter13=parameter13*parameter12 #cicj(Hij−ΔHmix)
parameter['Standard deviation of mixing enthalpy'] = parameter13.apply(lambda x: x.sum()**0.5, axis=1)
```
